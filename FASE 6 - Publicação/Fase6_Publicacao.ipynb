Resumindo o que precisa ser feito pra reproduzir o modelo do zero:

```
1. Carregar dados_governo.csv;
2. Limpar a coluna de remuneração (tirar "R$", ponto de milhar, 
   trocar vírgula por ponto e converter pra número);
3. Remover linhas com remuneração vazia ou igual a zero;
4. Calcular média e desvio padrão e aplicar o Box-Cox na remuneração;
5. Separar 80% treino e 20% teste (random_state=42);
6. Treinar o CatBoostRegressor usando o vínculo empregatício 
   como variável categórica;
7. Calcular RMSE, R² e a importância da variável.
```

Esse fluxo já está todo implementado nos arquivos `analise_exploratoria.py` (Check-in 01) e `modelagem.py` (Check-in 02) do repositório. Deixa explicito, que o `modelagem.py` conte codigos do `analise_exploratoria.py`, sendo uma continuação tendo a parte de CatBoost, que seria a parte de classificação e previsão, assim o Check-in 02 sendo o ultimo código a ser desenvolvido.

### 6.2.2 Saída

#### 6.2.2.1 Plano de Monitoramento e Manutenção

Como esse projeto não vai pra produção de verdade (é um trabalho da disciplina), o "monitoramento" aqui é mais no sentido de deixar claro o que precisaria ser cuidado se alguém continuar esse trabalho depois:

- **Atualização da base:** o arquivo `dados_governo.csv` reflete um momento específico da folha de pagamento da prefeitura. Se uma nova versão dos dados for baixada do portal da transparência, todo o pipeline de limpeza (Fase 3) precisa ser refeito, porque a proporção de registros `"null"` pode mudar de uma exportação pra outra.
- **Acompanhar o R² do modelo:** se esse modelo for reaproveitado em algum momento, vale anotar o R² obtido a cada nova rodada. Uma queda grande em relação ao 68.33% atual seria um sinal de que algo mudou na forma como os vínculos estão sendo registrados.
- **Resolver a questão do `"null"`:** antes de uma nova publicação, seria legal entender com a própria prefeitura por que mais da metade dos registros (53%) não tem vínculo identificado — pode ser um problema só de exportação dos dados, ou pode ser algo estrutural.

## 6.3 Relatório Final

### 6.3.1 Tarefa

Fechando o projeto, a gente escreve aqui um resumo geral de tudo que foi feito, dos resultados encontrados e das conclusões finais.

### 6.3.2 Saída

#### 6.3.2.1 Relatório Final

**Projeto:** Análise dos Gastos Públicos com Servidores da Prefeitura de Criciúma

**Pergunta-problema:** Como estão distribuídos os gastos públicos com servidores da prefeitura de Criciúma e quais padrões de alocação de recursos podem ser identificados, principalmente em relação ao tipo de vínculo empregatício?

**O que foi feito, resumindo cada fase:**

- **Fase 1 e 2** definiram o objetivo do projeto e mapearam a base de dados pública, com mais de 17 mil registros de servidores, contendo informações como cargo, vínculo, carga horária e remuneração.
- **Fase 3** cuidou da limpeza dos dados: convertemos a remuneração de texto pra número, removemos valores vazios e zerados, e sobraram 17604 registros válidos. A média salarial ficou em R$ 2015.45, com desvio padrão de R$ 2378.74, e cerca de 91.5% dos salários ficaram dentro de 1 desvio padrão da média (bem próximo dos 68% esperados numa curva normal). Como os testes de KS e Shapiro mostraram que a distribuição original não era normal, aplicamos a transformação Box-Cox, que reduziu bastante a assimetria dos dados (de aproximadamente 4,07 pra -0,09).
- **Fase 4** treinou um modelo CatBoost Regressor usando o vínculo empregatício como variável de entrada pra prever a remuneração (já transformada).
- **Fase 5** avaliou esse modelo: ele chegou a um R² de **0.6833** e RMSE de **0.3144**, confirmando que o vínculo é uma variável relevante pra explicar a remuneração, embora não seja a única. Também identificamos que 9253 registros (53% da base) estão com vínculo marcado como `"null"`, o que é uma limitação importante.
- **Fase 6** (esta) organiza a publicação de tudo isso.

**Principais achados:**

- A remuneração média dos servidores ficou em torno de R$ 2.015,45.
- O vínculo "Agente Político" tem a maior média salarial entre os tipos identificados (cerca de R$ 18.864), bem acima dos demais.
- "Contrato Administrativo Temporário" é o vínculo mais comum entre os registros identificados, com 5.216 ocorrências.
- O modelo confirma estatisticamente que o tipo de vínculo influencia a remuneração, com um poder de explicação de 68.33%.

**Limitações que ficam registradas:** o modelo usa só uma variável; não foi feita validação cruzada; mais da metade da base não tem o vínculo identificado corretamente; e mesmo depois do Box-Cox, os testes formais de normalidade continuam rejeitando a hipótese de distribuição normal (o que é esperado em bases tão grandes, mas precisa constar aqui).

## 6.4 Revisão do Projeto

### 6.4.1 Tarefa

Pra fechar, o grupo avalia o que deu certo, o que poderia ter sido feito diferente, e que aprendizado fica desse projeto.

### 6.4.2 Saída

#### 6.4.2.1 Documentação da Experiência

**O que funcionou bem:**

- Definir a pergunta-problema desde a Fase 1 ajudou o grupo a manter o foco em todas as fases seguintes, sem se perder em análises que não levavam a lugar nenhum.
- A limpeza da coluna de remuneração ficou bem resolvida e foi reaproveitada entre o Check-in 01 e o Check-in 02 sem precisar refazer nada.
- O CatBoost foi uma escolha boa porque lida direto com a variável categórica de vínculo, sem precisar fazer encoding manual.
- O Box-Cox funcionou bem pra reduzir a assimetria da remuneração, o que ajudou bastante na hora de treinar o modelo.
- O grupo conseguiu manter a documentação organizada fase por fase, o que facilitou bastante chegar até aqui sem perder o fio da meada.

**O que daria pra melhorar numa próxima vez:**

- O problema dos registros `"null"` no vínculo empregatício (mais da metade da base) deveria ter sido percebido já na Fase 2, quando a gente tava conhecendo os dados, e não só agora na Fase 5.
- O modelo ficou limitado a uma variável só. Dava pra ter testado outras colunas da base, como cargo ou carga horária, pra ver se o R² melhorava.
- Não rodamos validação cruzada, então as métricas de RMSE e R² dependem de um único jeito de separar treino e teste. Isso seria fácil de melhorar numa próxima rodada.
- Os testes de normalidade continuaram rejeitando os dados mesmo depois do Box-Cox. A gente justificou isso pelo tamanho da amostra, mas dava pra ter testado outras transformações (log, por exemplo) só pra comparar.

**O que o grupo aprendeu com isso:**

Esse projeto mostrou na prática como é importante checar a qualidade dos dados categóricos antes de jogar eles direto num modelo, e não confiar só no resultado final pra validar se está tudo certo. Seguir o CRISP-DM fase por fase ajudou a organizar o raciocínio, mas o aprendizado real foi perceber que mesmo um modelo "aprovado" pode (e deve) gerar uma lista de melhorias pra uma próxima versão do projeto.
