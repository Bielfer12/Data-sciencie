# Fase 1 - Entendimento do Negócio

Esta fase inicial foca no entendimento dos objetivos do projetos e requisitos de uma perspectiva de negócio, e então converter isso na definição do problema e de um plano preliminar.

## 1.1 Determinar os Objetivos do Negócio

### 1.1.2.1 Informações Internas
O projeto tem como base dados públicos disponibilizados pela prefeitura de Criciúma, especificamente relacionados ao quadro de funcionários e aos gastos com servidores. Esses dados são obtidos através do portal da transparência, garantindo acesso a informações atualizadas e confiáveis.

### 1.1.2.2 Objetivo de Negócio
O objetivo principal é analisar como estão distribuídos os gastos com servidores públicos da prefeitura de Criciúma, buscando entender a alocação de recursos e identificar possíveis padrões ao longo do tempo.

### 1.1.2.3 Critérios de Sucesso
O projeto será considerado bem-sucedido caso consiga:
- Identificar padrões relevantes nos gastos com servidores
- Gerar visualizações claras e compreensíveis
- Fornecer insights que ajudem na interpretação dos dados
- Possibilitar uma visão geral da evolução dos gastos ao longo do tempo

## 1.2 Avaliando a Situação

### 1.2.2.1 Inventário dos Recursos
- Dados públicos do portal da transparência da prefeitura de Criciúma
- Linguagem Python para análise
- Bibliotecas de análise de dados (como pandas, matplotlib, etc.)
- Ambiente de desenvolvimento (Jupyter Notebook ou similar)

### 1.2.2.2 Requisitos, Inferências e Condições
- Os dados devem estar acessíveis e em formato utilizável
- Pode ser necessário tratamento e limpeza dos dados
- Assume-se que os dados são confiáveis e refletem a realidade
- O projeto depende da qualidade e consistência das informações disponíveis

### 1.2.2.3 Riscos e Contingências
- Dados incompletos ou inconsistentes
- Dificuldade na interpretação de algumas informações
- Mudanças no formato dos dados ao longo do tempo
- Caso ocorram problemas, será necessário realizar ajustes no tratamento ou filtrar os dados

### 1.2.2.4 Terminologia
- Servidores públicos: funcionários vinculados à prefeitura
- Gastos: valores pagos a título de salários e benefícios
- Portal da transparência: plataforma pública de acesso aos dados
- Análise exploratória: etapa inicial de investigação dos dados

### 1.2.2.5 Custos e Benefícios

**Custos:**
- Tempo de desenvolvimento e análise
- Recursos computacionais

**Benefícios:**
- Maior transparência sobre o uso de recursos públicos
- Identificação de padrões e tendências
- Apoio à tomada de decisão baseada em dados

## 1.3 Determinar o Objetivo dos Modelos

### 1.3.2.1 Objetivo
Aplicar técnicas de análise de dados para explorar, visualizar e identificar padrões nos gastos com servidores públicos ao longo do tempo.

### 1.3.2.2 Critério de Sucesso
- Criação de gráficos e análises consistentes
- Identificação de tendências relevantes
- Capacidade de explicar os resultados de forma clara

## 1.4 Produzindo um Plano de Negócio

### 1.4.2.1 Plano de Projeto
O projeto será desenvolvido nas seguintes etapas:
1. Coleta dos dados no portal da transparência
2. Tratamento e limpeza dos dados
3. Análise exploratória
4. Criação de visualizações
5. Interpretação dos resultados

### 1.4.2.2 Avaliação Inicial de Técnicas e Ferramentas
Serão utilizadas ferramentas como Python e bibliotecas de análise de dados para manipulação e visualização. A análise exploratória será a principal técnica aplicada, podendo incluir gráficos, estatísticas descritivas e identificação de tendências.
